```python
from langchain_core.messages import HumanMessage, SystemMessage
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=1.0,
    max_tokens=None,
    max_retries=2,
)

# Read the resume file
with open("resume.txt", "r") as f:
    resume_content = f.read()

system_prompt = (
    "You are an expert resume analyst and career coach.\n"
    "Analyze the provided resume thoroughly and return a structured report covering:\n"
   
    "1. **Areas for Improvement** - Any gaps, missing sections, or weak points in the resume.\n"
    "2. **Overall Score** - Rate the resume out of 10 with a brief justification.\n"
    "Be concise, objective, and actionable in your feedback."
)

system_message = SystemMessage(content=system_prompt)

human_message = HumanMessage(
    content=f"Please analyze the following resume:\n\n{resume_content}"
)

response = model.invoke([system_message, human_message])
print(response.content)
```
